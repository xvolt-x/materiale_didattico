## 2. Il metodo del gradiente

Cercare m e q a occhio è lento. Il **metodo del gradiente** fa così: calcola in quale direzione l'errore **cresce** più rapidamente (il gradiente, cioè le derivate rispetto a m e a q) e fa un piccolo passo nella direzione **opposta**. Poi ripete.

Le due derivate dell'MSE sono:

$$\frac{\partial MSE}{\partial m} = -\frac{2}{n}\sum x_i\,(y_i - m x_i - q) \qquad \frac{\partial MSE}{\partial q} = -\frac{2}{n}\sum (y_i - m x_i - q)$$

e a ogni passo si aggiorna

$$m \leftarrow m - \eta\,\frac{\partial MSE}{\partial m} \qquad q \leftarrow q - \eta\,\frac{\partial MSE}{\partial q}$$

dove $\eta$ (eta) è la **lunghezza del passo**, in inglese *learning rate*.

```python
def passo(m, q, eta):
    scarti = y - (m * x + q)
    dm = -2 * np.mean(x * scarti)
    dq = -2 * np.mean(scarti)
    return m - eta * dm, q - eta * dq

m, q = 0.0, 0.0
for i in range(1, 2001):
    m, q = passo(m, q, eta=0.02)
    if i in (1, 10, 100, 500, 2000):
        print(f"passo {i:4d}:  m = {m:.4f}  q = {q:.4f}  MSE = {mse(m, q):.4f}")
```
